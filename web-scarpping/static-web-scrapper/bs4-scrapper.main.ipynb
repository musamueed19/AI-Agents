```python
import requests
from bs4 import BeautifulSoup

main_url = "toscrape.com"
sub_domain = "https://quotes." + main_url

response = requests.get(sub_domain)
page_html = response.text
soup = BeautifulSoup(page_html, 'html.parser')
```

```python
first_quote = soup.find('div', class_='quote')
first_quote
```

```python
quotes = soup.find_all('div', class_='quote')
len(quotes)
```

```python
for i, quote in enumerate(quotes):
    quote = quote.find('span', attrs={'itemprop': 'text'})
    if quote:
        print(i, quote.text)
    else:
        print(i, 'Quote Not Found!')
```

```python
tags_title = soup.find('h2', string='Top Ten tags')
tags_title
```

```python
# Navigation

Parent
Children
Sibling
```

```python
tag_box = tags_title.parent
tag_box.text
```

```python
first_tag_span = tags_title.find_next_sibling()
first_tag_span
```

```python
h2_tag = first_tag_span.find_previous_sibling()
h2_tag
```

```python
children = tag_box.children
children_list = list(children)
children_list = [child for child in children_list if child != '\n']
for i, child in enumerate(children_list):
    child_text = [child for child in list(child.text) if child != '\n']
    print(i+1, "".join(child_text[0:]))
    
```
